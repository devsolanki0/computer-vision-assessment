```python
import cv2
import numpy as np
from pathlib import Path

IMAGE_PATH = "food.jpg"
image = None

def load_image():
    global image
    image = cv2.imread(IMAGE_PATH)

    if image is None:
        print(f"Error: Could not read '{IMAGE_PATH}'.")
        return False

    h, w = image.shape[:2]
    channels = 1 if len(image.shape) == 2 else image.shape[2]
    print("Image loaded successfully.")
    print("Height:", h, "Width:", w, "Channels:", channels)

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    cv2.imshow("Original Image", image)
    cv2.imshow("Grayscale Image", gray)

    output = Path(IMAGE_PATH).stem + "_gray.jpg"
    cv2.imwrite(str(output), gray)
    print("Saved:", output)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def resize_analyse():
    if image is None:
        print("Please load the image first.")
        return

    resized = cv2.resize(image, (256, 256), interpolation=cv2.INTER_AREA)
    b, g, r = cv2.split(resized)

    print("Blue average:", round(float(np.mean(b)), 2))
    print("Green average:", round(float(np.mean(g)), 2))
    print("Red average:", round(float(np.mean(r)), 2))

    cv2.imshow("Resized 256x256", resized)
    cv2.imshow("Blue Channel", b)
    cv2.imshow("Green Channel", g)
    cv2.imshow("Red Channel", r)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def transformation_pipeline():
    if image is None:
        print("Please load the image first.")
        return

    angle = float(input("Enter rotation angle in degrees: "))
    h, w = image.shape[:2]
    matrix = cv2.getRotationMatrix2D((w//2, h//2), angle, 1.0)
    rotated = cv2.warpAffine(image, matrix, (w, h))

    rh, rw = rotated.shape[:2]
    crop_h, crop_w = int(rh * 0.60), int(rw * 0.60)
    y1 = (rh - crop_h) // 2
    x1 = (rw - crop_w) // 2
    cropped = rotated[y1:y1+crop_h, x1:x1+crop_w]

    flipped = cv2.flip(cropped, 1)

    cv2.imshow("1 - Rotated", rotated)
    cv2.imshow("2 - Central 60% Crop", cropped)
    cv2.imshow("3 - Horizontal Flip", flipped)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def edge_quality_scan():
    if image is None:
        print("Please load the image first.")
        return

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    edges = cv2.Canny(blurred, 80, 160)

    edge_pixels = cv2.countNonZero(edges)
    print("Total edge pixels:", edge_pixels)

    if edge_pixels > 5000:
        print("Quality verdict: High texture (good detail)")
    else:
        print("Quality verdict: Low texture (may need re-shoot)")

    cv2.imshow("Edge-Based Quality Scan", edges)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

while True:
    print("\n===== Food Delivery Image Quality Inspector =====")
    print("1. Load & inspect image")
    print("2. Resize and analyse colour channels")
    print("3. Apply transformation pipeline")
    print("4. Run edge-based quality scan")
    print("5. Exit")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        load_image()
    elif choice == "2":
        resize_analyse()
    elif choice == "3":
        transformation_pipeline()
    elif choice == "4":
        edge_quality_scan()
    elif choice == "5":
        print("Program ended.")
        break
    else:
        print("Invalid choice. Please select 1 to 5.")
```
