```python
# 1

import cv2

image = cv2.imread("food.jpg")
resized = cv2.resize(image, (300, 300))
gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 0)
edges = cv2.Canny(blur, 80, 160)

edge_pixels = cv2.countNonZero(edges)
print("Edge pixels:", edge_pixels)

combined = cv2.hconcat([resized, cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)])
cv2.imshow("AI Edge Detection", combined)

cv2.imwrite("ai_edge_output.jpg", edges)
print("Saved ai_edge_output.jpg")

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 2

import cv2
import sys

IMAGE_PATH = "food.jpg"

image = cv2.imread(IMAGE_PATH)

if image is None:
    print(f"Error: Could not read '{IMAGE_PATH}'. Check the filename/path.")
    sys.exit()

resized = cv2.resize(image, (300, 300))
gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 0)
edges = cv2.Canny(blur, 80, 160)

edge_pixels = cv2.countNonZero(edges)
print("Total edge pixels:", edge_pixels)

edge_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
combined = cv2.hconcat([resized, edge_bgr])

cv2.imshow("Food Image and Edge Map", combined)

saved = cv2.imwrite("ai_edge_output.jpg", edges)
if saved:
    print(f"Saved: ai_edge_output.jpg | Edge pixels: {edge_pixels}")
else:
    print("Error: Could not save ai_edge_output.jpg")

cv2.waitKey(0)
cv2.destroyAllWindows()
```
