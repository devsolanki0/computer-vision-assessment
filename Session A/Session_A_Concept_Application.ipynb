### Task 1 — Computer Vision vs Rules-Based Filtering
Computer Vision is a field of AI that enables computers to understand and analyse images or video. A rules-based filter depends on fixed conditions such as colour or brightness thresholds. CV is more suitable for diverse food images because food can have different colours, shapes, textures and lighting conditions. A trained CV model can learn useful visual patterns instead of depending only on fixed thresholds.

### Task 2 — Image Preprocessing
First read the image, check that it loaded correctly, resize it to the model input size (128×128), and keep the expected number of channels. RGB images should remain RGB when colour information is important. Grayscale can be useful when shape, edges or intensity are enough and it reduces computation. For freshness detection, grayscale can lose colour information such as browning, redness, greenish changes or other colour cues, so RGB should be retained when those cues matter.

### Task 3 — Rotation, Cropping and Flipping
The image should first be rotated to correct its orientation. Then crop the lower-right region because the label location is defined after orientation is corrected. Finally flip horizontally only if required by the label orientation. Cropping before rotation can select the wrong physical region because the lower-right quadrant may change after rotation.

### Task 4 — Canny Edge Detection
Canny first uses image gradients to find strong intensity changes. Non-maximum suppression keeps thin edge responses, and hysteresis uses low and high thresholds to connect strong edges with relevant weaker edges. Gaussian blur is applied before Canny to reduce noise and small unwanted edges. A higher high threshold gives fewer strong edges, while a lower threshold can detect more weak edges but may also add noise. Thresholds should be tuned so char marks are detected without detecting every plate rim.

### Task 5 — Haar Cascade False Positives
False positives can occur when image patterns resemble the features learned by the Haar cascade. `scaleFactor` controls the image pyramid step size, `minNeighbors` controls how many nearby detections are needed to keep a detection, and `minSize` rejects faces smaller than the selected size. Increasing `minNeighbors` usually reduces false positives but can remove real faces. For example, if a delivery agent is far from the camera and the face is small or partly hidden, a very high `minNeighbors` may cause the real face not to be detected.

### Task 6 — Video Frame Sampling
Processing and saving every frame increases CPU usage, storage and latency. A fixed sampling interval processes fewer frames and is suitable when every frame is not necessary. For important motion, a practical strategy is to sample at a reasonable interval and use motion/event detection between samples so important activity is not missed. `cap.set(cv2.CAP_PROP_POS_FRAMES, n)` moves the video position to frame `n`, which can be used to implement controlled frame skipping.
