## Task 1 — Food Image Loader and Inspector

```python
import cv2
import sys

IMAGE_PATH = "food.jpg"

image = cv2.imread(IMAGE_PATH)

if image is None:
    print(f"Error: Could not read '{IMAGE_PATH}'. Check the filename/path.")
    sys.exit()

height, width, channels = image.shape
print("Image loaded successfully.")
print("Height:", height)
print("Width:", width)
print("Colour channels:", channels)

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imshow("Original Food Image", image)
cv2.imshow("Grayscale Food Image", gray)

cv2.imwrite("grayscale_food.jpg", gray)
print("Saved grayscale image as: grayscale_food.jpg")

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Task 2 — Menu Photo Resizer and Colour Channel Analyser

```python
import cv2
import numpy as np

IMAGE_PATH = "food.jpg"
image = cv2.imread(IMAGE_PATH)

if image is None:
    raise FileNotFoundError(f"Could not read: {IMAGE_PATH}")

resized = cv2.resize(image, (256, 256), interpolation=cv2.INTER_AREA)
blue, green, red = cv2.split(resized)

print("Average Blue intensity:", round(float(np.mean(blue)), 2))
print("Average Green intensity:", round(float(np.mean(green)), 2))
print("Average Red intensity:", round(float(np.mean(red)), 2))

averages = {"Blue": np.mean(blue), "Green": np.mean(green), "Red": np.mean(red)}
highest = max(averages, key=averages.get)
print(f"Highest average channel: {highest}")
print(f"This suggests that {highest} is the dominant colour component in the dish.")

cv2.imshow("Resized 256x256 Image", resized)
cv2.imshow("Blue Channel", blue)
cv2.imshow("Green Channel", green)
cv2.imshow("Red Channel", red)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Task 3 — Packaging Label Transformer and Edge Scanner

```python
import cv2

IMAGE_PATH = "packaging.jpg"
image = cv2.imread(IMAGE_PATH)

if image is None:
    raise FileNotFoundError(f"Could not read: {IMAGE_PATH}")

height, width = image.shape[:2]
center = (width // 2, height // 2)

matrix = cv2.getRotationMatrix2D(center, -90, 1.0)
rotated = cv2.warpAffine(image, matrix, (width, height))

rh, rw = rotated.shape[:2]
cropped = rotated[rh // 2:rh, rw // 2:rw]

flipped = cv2.flip(cropped, 1)

blurred = cv2.GaussianBlur(flipped, (5, 5), 0)
edges = cv2.Canny(blurred, 80, 160)

cv2.imshow("Rotated Image", rotated)
cv2.imshow("Cropped Lower-Right Quadrant", cropped)
cv2.imshow("Flipped Image", flipped)
cv2.imshow("Label Edges", edges)

cv2.imwrite("label_edges.jpg", edges)
print("Saved edge map as: label_edges.jpg")

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Task 4 — Live Delivery Agent Face Verifier

```python
import cv2

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open the webcam.")
else:
    while True:
        ret, frame = cap.read()

        if not ret:
            print("Error: Could not read webcam frame.")
            break

        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        faces = face_cascade.detectMultiScale(
            gray,
            scaleFactor=1.1,
            minNeighbors=6,
            minSize=(80, 80)
        )

        for (x, y, w, h) in faces:
            cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
            cv2.putText(
                frame, "Agent Detected", (x, y-10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2
            )

        cv2.imshow("Delivery Agent Verifier", frame)
        key = cv2.waitKey(1) & 0xFF

        if key == ord("s"):
            cv2.imwrite("agent_snapshot.jpg", frame)
            print("Snapshot saved as: agent_snapshot.jpg")

        elif key == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()
```
